```python
import numpy as np

def stable_softmax(x: np.ndarray, axis: int = -1) -> np.ndarray:
    """
    Computes numerical stable softmax along the specified axis.
    
    Formula: exp(x_i - max(x)) / sum(exp(x_j - max(x)))
    """
    max_x = np.max(x, axis=axis, keepdims=True)
    print(max_x)
    print(x - max_x) # max_x broadcast to fit same size as x

    exp_x = np.exp(x - max_x)
    print(exp_x)
    print(np.sum(exp_x, axis=axis, keepdims=True)) # sum broadcast to fit same size as exp_x

    return exp_x / np.sum(exp_x, axis=axis, keepdims=True)
```

```python
arr1 = np.array([[1, 2, 3, 4], [5, 16, 7, 28]]) # 2D array
print(stable_softmax(arr1))
```

```python
arr2 = np.array([
    [[1, 2, 3, 4],
     [5, 16, 7, 28]],
    [[2, 4, 6, 8],
     [3, 4, 2, 9]]]) # 3D array
print(stable_softmax(arr2))
```

```python
print(arr2 / 5) # broadcasting
```
